```python
import pandas as pd

df = pd.read_csv("raw_retail_sales_12000.csv")

print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

display(df.head())
```

```python
clean_df = df.copy()

# Standardize column names
clean_df.columns = (
    clean_df.columns.str.strip().str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
)

# Remove exact duplicate records
duplicates_removed = clean_df.duplicated().sum()
clean_df = clean_df.drop_duplicates().copy()

# Convert quantity and price columns to numeric
for col in ["quantity", "unit_price", "line_total"]:
    clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

# Convert dates, handling mixed formats
clean_df["datetime"] = pd.to_datetime(
    clean_df["datetime"], format="mixed", errors="coerce"
)

# Fill missing numeric values with medians
for col in ["quantity", "unit_price", "line_total"]:
    clean_df[col] = clean_df[col].fillna(clean_df[col].median())

# Fill missing categorical values
for col in ["store_id", "product", "department", "payment"]:
    clean_df[col] = clean_df[col].fillna("Unknown")

# Fill missing dates using the most frequent valid date
if clean_df["datetime"].notna().any():
    clean_df["datetime"] = clean_df["datetime"].fillna(
        clean_df["datetime"].mode().iloc[0]
    )

# Create date features
clean_df["year"] = clean_df["datetime"].dt.year
clean_df["month"] = clean_df["datetime"].dt.month
clean_df["month_name"] = clean_df["datetime"].dt.month_name()

print("Duplicates removed:", duplicates_removed)
print("Remaining missing values:", clean_df.isnull().sum().sum())
print("Final shape:", clean_df.shape)

display(clean_df.head())
```

```python
clean_df.to_csv("clean_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

before_report = {
    "Rows": len(df),
    "Columns": len(df.columns),
    "Missing Values": int(df.isnull().sum().sum()),
    "Duplicate Rows": int(df.duplicated().sum())
}

after_report = {
    "Rows": len(clean_df),
    "Columns": len(clean_df.columns),
    "Missing Values": int(clean_df.isnull().sum().sum()),
    "Duplicate Rows": int(clean_df.duplicated().sum())
}

summary = pd.DataFrame({
    "Metric": before_report.keys(),
    "Before Cleaning": before_report.values(),
    "After Cleaning": after_report.values()
})

display(summary)

summary.to_csv("cleaning_summary.csv", index=False)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Missing values comparison
axes[0].bar(
    ["Before", "After"],
    [before_report["Missing Values"],
     after_report["Missing Values"]]
)
axes[0].set_title("Missing Values Before vs After")
axes[0].set_ylabel("Count")

# Duplicate rows comparison
axes[1].bar(
    ["Before", "After"],
    [before_report["Duplicate Rows"],
     after_report["Duplicate Rows"]]
)
axes[1].set_title("Duplicate Rows Before vs After")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.savefig("cleaning_comparison.png", dpi=150)
plt.show()
```
